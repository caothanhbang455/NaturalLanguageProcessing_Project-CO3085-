```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
documents = [
    'I love machine learning.',
    'Machine learning is powerful.',
    'Python is a programming language.',
    'NLP is Natural Language Processing.'
]
```

```python
vectorizer = CountVectorizer()
```

```python
X = vectorizer.fit_transform(documents)
```

```python
cosine_similarity(X, X)
```

```python
user_input = 'I\'m happy to learn machine learning.'
```

```python
y = vectorizer.transform([user_input])
```

```python
matching_result = cosine_similarity(y, X)
matching_result
```

```python
matching_result[0]
```

```python
sort_result = np.argsort(matching_result[0])[::-1]
sort_result
```

```python
top_doc = [documents[i] for i in sort_result]
top_score = [matching_result[0][i] for i in sort_result]
```

```python
for doc, score in zip(top_doc, top_score):
    print(f'Document={doc} have similar to user_input with score={score}')
```
